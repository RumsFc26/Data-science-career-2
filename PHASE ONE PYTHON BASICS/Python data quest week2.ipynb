```python
#TASK 8 :Word Frequency Counter: Write a script that reads a .txt file (your choice — any article or text), counts the frequency of every word, 
# and prints the top 15 most common words with their counts. Ignore common words like "the", "a", "is". 



#word frequency counter
with open("article.txt","r") as file :
    content =  file.read()
    words = content.lower().split()
    common_words = ["this","a","is"]
    word_count= {}
for word in words :
        if word in common_words:
          continue
        elif word in word_count:
            word_count[word] = word_count[word] + 1
        else:
            word_count[word] = 1
            
 # note that the outputs from the for loop above are used to fill the dictionary called (word_count)
 #items() will convert the dictionary in to a list of key value pairs of the form ("word",frequency) which is now a tuple,we the sorted the tuple to see the word with highest frequency

sorted_words = sorted(word_count.items(), key = lambda x: x[1], reverse=True)
#lambda here is a function that takes an agument such as a  key from the  list and it returns the value of that key which is the frequency of the word in this case,
#and reverse  + true will sorted the list and arrange the values from biggest to smallest instead of the normal sorting that we know which is from smallest to biggest

# now we can print the top ten most common words in the article from the sorted_words list(sorted_words is a list of tuples where each tuple is in the form ("word",frequency))

# now we loop through the sorted list amd print the top 15 most common words together with their counts
for word,count in sorted_words[:15]:
     print(word, ":",count)
#note tat we could also sort using the bubble sort try to implement this        
    


```

```python
#TASK 10: error handler Error Handler: Adapt your CSV parser to handle: missing files (FileNotFoundError), empty rows,
#and non-numeric values in the score column. The script should log 
#errors and continue processing rather than crashing.

filename = "students.csv"

# 1. Creating  the  CSV file  but this time with some invalid data from which we will manage errors
students_info = [
    ["Name", "Age", "Score", "Grade"],
    ["rums", 21, 71, "B"],
    [],
    ["anna", 19, 85, "A"],                      
    ["oscar", 22, "fifty-five", "C"],    
    ["clement", 20, 90, "A"],
    ["micheal", 18, 55, "D"],
    ["brandon", 22, 70, "B"],
    ["Nelson", 23, 88, "A"],
    ["Harry", 20, 40, "F"],
    ["Benita", 19, 98, "A"],
    ["jackson", 18, 65, "C"]
]
#2. Writing on the csv file
with open(filename, "w") as file:
    for row in students_info:
        if len(row) == 0:
            file.write("\n")
        else:
            line = ",".join(map(str, row))
            file.write(line + "\n")
scores = []
students_data = []
err_log = []

# Error handling block
try:
    with open(filename, "r") as f:
        content = f.readlines()

        # Skiping  the header
        for line_num, line in enumerate(content[1:], start=2):
            # Checking  for an empty row
            if line.strip() == "":
                err_messg = "Line " + str(line_num) + ": Empty row"
                err_log.append(err_messg)
                continue
            parts = line.strip().split(",")

            # Checking  if  the row has enough columns
            if len(parts) < 4:
                err_messg = "Line " + str(line_num) + ": Incomplete row"
                err_log.append(err_messg)
                continue
            name = parts[0]

            # Converting the  age
            try:
                age = int(parts[1])
            except ValueError:
                err_messg = (
                    "Line " + str(line_num) +
                    ": Invalid age for " + name
                )
                err_log.append(err_messg)
                continue

            # Converting the score
            try:
                score = int(parts[2])
            except ValueError:
                err_messg = (
                    "Line " + str(line_num) +
                    ": Non-numeric score for " + name
                )
                err_log.append(err_messg)
                continue
            grade = parts[3]
            
            # Adding a  valid  student  data
            students_data.append((name, age, score, grade))
            scores.append(score)

# 3. Message for Handling  missing file
except FileNotFoundError:
    err_messg = "Error: The file '" + filename + "' was not found."
    err_log.append(err_messg)

# 4. Calculating the average score
if len(scores) > 0:
    avg_score= sum(scores) / len(scores)
    print("Average score is:", avg_score)
    print("\nStudents who scored above average:")
    for student in students_data:
        if student[2] > avg_score:
            print(student[0])

else:
    print("No valid scores were found.")
# 5. Displaying  the  error log
print("\nERROR LOG:")

if len(err_log) == 0:
    print("No errors found.")
else:
    for error in err_log:
        print(error)
```
