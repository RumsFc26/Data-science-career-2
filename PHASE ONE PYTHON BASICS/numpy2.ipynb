```python
#Numpy Random number Generation
# Random Numbers here simply means numbers that are not predictable(every time you run the code the numbers keep changing) 
import numpy as np
#1 Random floats 
print("1D random floats:",np.random.rand(3))
print("2D random floats:\n",np.random.rand(2,2))
#2 Random integers(generate random integers within a specific range)
print("a single random integer:",np.random.randint(1,7))
print("array of random integers:\n",np.random.randint(1,10,5))
# the 5 represents the size ie a 1D array of size = 5 elements and (5,5 means 25 elements) 
#3 Distributions(we will see this in statistics ,suppose we want to generate numbers within some distribution)
#randn() : generates samples from the standard normal distribution(that is a D of mean = 0 and std = 1 )
#normal() : allows you to costumize mean and standard deviation
#uniform() : gives values with a particular or specific  uniform range
print("Standard Normal:",np.random.randn(3)) 
print("Normal Distribution(μ = 4,σ = 1):",np.random.normal(loc = 0, scale = 1,size = 5))
print("Uniform Dist (5,15):",np.random.uniform(5,15, size = 3))

```

```python
#Concept of reproducibility with seed
# here we want to generate numbers that will not bet change as you run the code 
np.random.seed(42)
# you could put any number in brackets 
print("Standard Normal:",np.random.randn(3)) 
```

```python
#Concept of shufflling and sampling
# choice():selects random elements from an array
# shuffle() :randomly rearranges elements in place 
data = [10, 20, 30, 40, 50] 
print("random sample(no replacement):",np.random.choice(data, size = 3,replace = False))
np.random.shuffle(data)
print("shuffled array:",data)
```

```python
#Reshaping means changing the shape(dimension) without changing the data
# suppose you have a 1D and you want to reshape to a 2D do this
u = np.array([1,2,3,4,5,6])
u
print("reshaped :",u.reshape(2,3))
```
