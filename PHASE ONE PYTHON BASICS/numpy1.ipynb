```python
import numpy as np
#Numpy is the fundamental package for numerical computing in python.it support for arrays, matrices and collection of mathematical functions to operate on these data structures 
#WHY USE NUMPY
#1,Efficient storage and manipulations of numerical arrays
#2,Faster computations using Vectorization
#3,Essential for scientific computations,Data science and Machine learning    
```

```python
#creating arrays using the numpy library
# to declare an array using numpy, you will use the prototype np.array()then you pass a list inside ie n.p.array([])
#Creating 1 dimensional array(it is like a single list of elements in one line(one row)) 
import numpy as np
a = np.array([1,2,3,4])
b = np.array([4,5,6,1])
print("1D array a is :",a)
print("1D array b is :",b)
#get the data type of an array
print(type(b))
#get the data type of a value stored in an array
print(type(b[0]))
# you could also assign a data type to the values of your own array as shown below
b = np.array([4,5,6,1],dtype = np.int8)
print(type(b[2]))
# you could pass in a string and you convert it in to a float or any other  dtype  automatically  as shown below. note this will only work if the string is unique to the elements of the list and not something strange as show
b3 = np.array([4,5,6,"1"], dtype = np.float16)
print(type(b3[3]))


```

SIMPLY 1D ARRAYS IS A SINGLE LIST WHILE 2D IS A NESTED LIST

```python
#Creating multi dimensional arrays(it is an array that contains other arrays as its elements,instead of containing one line (1 dimension) the data is stored in multiple dimensions that is in rows and columns)
c = np.array([[1,2],[3,4]])
print("2D array:",c)
```

```python
#some built in functions for Numpy
#1 this will generate a matrix made up of zeroes with 2 rows and 3 columns. IT IS ACTUALLY A 2D MATRIX BUT HAS 2 ROWS AND 3 COLUMNS 
zeros =  np.zeros((2,3))
print(zeros)
#2 this built in function can be used generate and array made up of ones
ones = np.ones((3,3))
print(ones)
##Note that when we use print,it is actually the content of the array that is displayed to us but when we just write the variable at the end of our code,the created array is shown to us like below
#3 
d = np.arange(10,50)
print(d)
d
#4 line space is used to generate numbers with evenly spaced values line space(start,stop,number of values)
sp = np.linspace(0,1,5)
sp
print(sp)
```

```python
#ARRAY ATTRIBUTES(some are : shape,size,dtpye,ndim)
# you could determine the shape of an array and you could also give a shape to any array you created 
# lets determine the shape of the above arrays
print("shape is :",c.shape)
print("shape is :",a.shape)
print("size is :",a.size)
print("Number of dimensions :",c.ndim)
print("data type of b is :",b.dtype)

 
```

```python
#slicing and indexing 
# note that counting of rows and columns begins from zero so we have row zero the row one etc
print("element at position (1,1) fom c is :",c[0,0])
print("get the first row :",c[0])
print("last column is :",c[:,-1])
print("first element of 1D array is :",a[0])
print("last element of 1D array is :" ,a[-1])
```

```python
#Mathematical operations on arrays
#we can only add and subtract two vectors having the same dimensions 
print("Addition :",a + b)
print("Subtraction :",a - b)
print("Exponential:",a ** 2)
print("sin :",np.sin(a))
#suppose we want to plot a graph of sin a against sign b this will be very useful
print("Dot product:",np.dot(a,b))
```

```python
#Aggregation functions
print("average or mean of c is :",np.mean(c)) 
print("maximum  element of a is :",np.max(a))
print("minimum element of b is :",np.min(b))
print("sum of c is :",np.sum(c))
# say you want to look for the minimum or maximum value following some restrictions like row-wise or column_wise 
print("max element of c column wise is :",np.max(c,axis = 0))
print("max element of c row wise is :",np.max(c,axis = 1))
print("min element of c row wise is :",np.min(c,axis = 1))
print("min element of c column wise is :",np.min(c,axis = 0))
```

```python
#Broadcasting is a powerful  feature in NUMPY that allows arithmetic operations between arrays of different shapes without explicitly copying and reshaping the array
x = np.array([[1,2,3],[4,5,6]]) 
y = np.array([10,20,30]) 
print(x)
print(y)
print("Broadcast is :",x + y)


#sorting and filtering 
#1 sorting helps us to arrange elements either in alphabetical order,ascending or descending order
numbers = np.array([9,8,7,6,5,4,3,2,1,0])
fruits  = np.array(["banana","apple","cherry"])
print("sorted numbers are :",np.sort(numbers ))
print("sorted numbers fruits are :",np.sort(fruits))
# filtering also means selecting data based on a particular condition
mask = numbers > 3
print("filtered numbers : ",numbers[mask])

```
