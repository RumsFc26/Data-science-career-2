```python
#task two:a created dictionary of 8 african countries with their GDP per capita(USD).a written for loop that categorises each as low income(<1,000),Middle income (100 to 12000) or High income(>12000) and a printed summary for each
countries = {"cameroon":1000,"nigeria": 500,"south africa": 2000,"egypt":5000, "kenya": 700,"congo":15000, "gabon":7500,"algeria":12000}
for country in countries:
    if countries[country] < 1000:
        print(f"the  country {country}is a low income country of gpd {countries[country]}")
    elif countries[country] >= 1000 and countries[country] <= 12000:
        print(f"the country{country} is a middle income country with gfd of{ countries[country]}")
    else :
        print(f"the country {country} is a high income count with gdp of {countries[country]}")   

```
