```python
with open("dog_breeds.txt","r") as f1:
  line = f1.readline()
  while line != " ":
    print(line, end =" ")
    line = f1.readline()
```

```python
#you want to read from a file and write to another fille at the same time
d_path = 'demo.text'
d_r_path = 'dog_breeds.txt'
with open(d_path, 'r') as reader, open(d_r_path, 'w') as writer:
    dog_breeds = reader.readlines()
    writer.writelines("Rumsphyld would like to be a data scientist and a machine learning engineer")
    print(dog_breeds)

with open("dog_breeds.txt","r") as reader :
    print(reader.read())
```
