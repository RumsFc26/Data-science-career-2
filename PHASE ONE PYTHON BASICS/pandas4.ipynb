# Conditional Selection

```python
# get all the rows where a persons age is greater than 70
# use format : df[condition]
# conditional statements are also used to slice or filter the data frame based on some conditions 
# Normally  a conditional statement is used to make a decision based on a condition and this decision is made using the result of a comparison operator which is boolean(either true or false) as show below.
df["Age"]>70
```

```python
# Now to get all the rows where a persons age is greater than 70, only the rows for which the above condition is true will be returned 
df[df["Age"]>70]
# You could store it in a variable called older_passengers = df[df["Age"]>70] and do older_passengers.head()
```

```python
# suppose we have multiple conditions like get all the females who survived
df[(df["Sex"] == "female")&(df["Survived"]==1)]
```

```python
# Suppose you want to use this condition but to get only Name and Age information and also get for only the first 5 rows that satisfy the condition
female_survivors_subset = df[(df["Sex"] == "female")&(df["Survived"]==1)][["Name", "Age"]]
female_survivors_subset.head()
# only & is compatible with pandas and not(and or &&)
```

```python
#1 get all the people who survived 
#df["Survived"].value_counts()
#OR
# To be more direct, we use this one below as it give us only survivors and says nothing about non survivors like the other one
Survived = len(df[df["Survived"] == 1])
print(Survived)
#2 get all people who did not survived
not_Survived = len(df[df["Survived"] == 0])
print(not_Survived)


```

```python
#percentage of:(this percentage we got from the data is an example of what we calls data insights)
#1 people who survived
print(Survived/(Survived+not_Survived))
#2 people who did not survived
print(not_Survived/(Survived+not_Survived))
```

```python
#Another way of getting all survivors who are female and survived is using the loc(survivors subset).
survivors_subset_loc = df.loc[((df["Sex"] == "female")&(df["Survived"]==1)), ["Name", "Age"]]
survivors_subset_loc.head()
#If you notice, here we are passing this: ["Name", "Age"] instead of this :[["Name", "Age"]] when we were using df to  get all survivors who are female and survived above.here we use this: ["Name", "Age"] because loc is like a function and takes a list as an input compared to df which follows the [["Name", "Age"]] format.
```

```python
#when you want to carry out some operations on the data frame, always try to use the ".copy()" method to clone the data frame before slicing so  the original data frame is not affected
#Note here that selection using loc will may a view or reference of the df where by if any change is applied to this view, it affects the original df.
#Note also that selecting with iloc gives subset which is from a copy of the df where by if any change is applied to this copy, it does not affects the original df.

#lets trying using loc and see
#we will get the error message so preferable we use iloc or best approach is to use the copy method to avoid any forms of error completely
subset_male = df.loc[df["Sex"] == "male"]
subset_male["Age"] = subset_male["Age"] + 1
print(subset_male)
```

```python
# This is the best way when we want to do any modification we make a clone copy of the original df first and do the modifications on it so that our original dat is not modified or interrupted.
subset_male = df.loc[df["Sex"] == "male"].copy()
subset_male["Age"] = subset_male["Age"] + 1
print(subset_male)
# we can check the official documentation for pandas on google for details on copy and view
#using simple subset = df[condition] might create a copy or a view but to be sure that it is a copy as we require, always attach the ".copy()" method to the condition ie subset = df[condition].copy()
```

```python
#Notes
# iloc ==> creates a copy of df
# loc ==> may create a view
# df[...] ==> it also might be view
# always  use .copy() to be safe
# the reason for creating a copy and working with it is because when you create a copy, and why doing you analysis, you encounter a problem along the way, you will be able to revert to the original data frame.
```

# Reindexing 
it helps align or make two columns to be equal or compatible so that we can perform some operations on them

```python
#looking at a scenario or a real case where we have stores and sales in the month of january and the month of february.we you look clearly in the month of february, store B closed. 
sales_jan = pd.DataFrame({"Sales":[250, 400, 300]}, index=["store_A","store_B","store_C"])
sales_jan
```

```python
Sales_feb = pd.DataFrame({"Sales":[260,310]},index = ["store_A","store_C"])

Sales_feb
```

```python
# Now suppose we want to check which sales month has the more stable indexes(the more stable index is the one with more rows or more index rows)
common_index = sales_jan.index
print(common_index)
```

```python
# Now, we use "reindex" to reindex sales_feb which has unstable column(lesser number of rows) using the indexes for sales_jan(we make it take the format for sales_jan) 
sales_feb_aligned = Sales_feb.reindex(common_index, fill_value = 0)
# Normally, we are to have :Sales_feb.reindex(common_index) but since we want to do calculations with the data, we have to use fill_value = 0 to fill empty rows so that calculations will e sensible
sales_feb_aligned
```

```python
 #Now we can start applying operations on the data since both sales_jan columns and sales_feb columns are now in the same format applying operations on them eg sum, difference
sales_diff = sales_feb_aligned - sales_jan
sales_diff
```

```python
# now we need to add store  D for the common index that has to be used to reindex A and B to be complete ie(A,B,C,D). we can do this by adding another store to sales feb for it to be in the required format by the question(A,C and D)
Sales_feb = pd.DataFrame({"Sales":[260,310,100]},index = ["store_A","store_C","store_D"])

Sales_feb
```

```python
# now suppose we have a case where sales_jan is A,B and C while sales_feb is A,C AND D,IN THIS CASE WE WILL HAVE TO REINDEX BOTH sales_jan AND sales_feb BEFORE performing any operations on them. we will actually need to create a new common index having complete indexes A,B,C and D to reindex sales_jan and sales_feb 
common_index = list(set(list(sales_jan.index) + list(Sales_feb.index)))
print(common_index)

```

```python
# lastly reindexing both sales_jan and sales_feb so that we can perform operations on them. this is shown below
sales_jan_aligned =  sales_jan.reindex(common_index,fill_value = 0)
sales_jan_aligned
# know that what is returned is a df so you can execute it like a  df
```

```python
# reindexing sales_feb
# since we reindexed it before we shall now use a new variable ie sales_jan_aligned_new
sales_feb_aligned_new =  Sales_feb.reindex(common_index,fill_value = 0)
sales_feb_aligned_new
```

```python
# you then fill empty values and perform your operations. fill empty values with fill_value = 0

#now lets do an operation like selecting rows of both sales_jan and sales_feb_new
sales_feb_aligned_new.iloc[1]
sales_jan_aligned.loc["store_C"]
#etc
```
