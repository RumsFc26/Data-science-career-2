Date Functionality in Pandas

```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "event": ["concert", "conference", "wedding", "reception","last_day"],
    "date":["2025-01-01", "2025-03-15","2025-07-20","2025-07-21","2025-07-31"]
})
```

```python
# looking at the data type of the date_column
print(df["date"].dtype)
```

```python
# Now suppose we want  to convert this date column to a new type called "datetime"
df["date_pandas"] = pd.to_datetime(df["date"])
df
```

```python
# Now checking the dtype of date_pandas
print(df["date_pandas"].dtype)
```

```python
# looking closely, the columns("date"&"date_time") look same but if we check the data types, they look different, lest verify
print(df.dtypes[["event", "date", "date_pandas"]])
# we will learn more about date time in ML
```

```python
#Now that the column is of date time, we can explore many things 
#1 you can check the year
df["date_pandas"].dt.year
```

```python
#2, you can check the month
df["date_pandas"].dt.month
```

```python
#3, you can check the day
df["date_pandas"].dt.day
```

```python
#4 you can check the day of the week
df["date_pandas"].dt.weekday
# two(2 here means second day of the week) or second day of a particular week
# dt simply means date same as we were using str for string
```

```python
# you data should only have one format, if your data has two formats, the your data is broken
# YYYY-MM-DD
# YYYY-DD-MM
# only one format and not both
```

```python
df["date_pandas"].dt.day_name()
# day number starts from zero and end at 6":
# 0 - Monday
# 6 - Sunday
# never you change this logic of  pandas 
```

```python
df["date_pandas"].dt.month_name()
```

```python
# checking if a particular day is the month end or not
df["date_pandas"].dt.is_month_end
```

```python
# check events that are yet to come between two dates 
df[df["date_pandas"].between("2025-01-01", "2025-06-01")]
```

```python
#check the dates or events after a particular date
df[df["date_pandas"] > "2025-06-01"] 
```

```python
# Date Range
pd.date_range(start ="2024-01-01", end = "2024-01-10", freq = "D")
# check the number o dates(days) between two dates 
```

```python
# checking  the dates which a month ends in a column for dates 
# simply swap between freq to explore more ie use "ME" for month end, "MS" for month starts 
pd.date_range(start ="2024-01-01", end = "2024-03-10", freq = "ME")
```

# Date Arithmetic

```python
# Lets say we want to add 7 days to every date to get the next week
df["next_week"] = df["date_pandas"] + pd.Timedelta(days = 7)
```

```python
df[["date_pandas", "next_week"]]
```

```python

df["days_to_event"] = df["date_pandas"] - pd.Timestamp.today()
df
```
