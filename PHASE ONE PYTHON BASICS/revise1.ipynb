```python
n = 31
for side1 in range(1,n):
    for side2 in range((side1+1),(n + 1),1):
        sum_of_squares_of_two_sides = (side1*side1)+(side2*side2)
        for longest_side in range((side1+2),(n +1),1):
            if(longest_side == sum_of_squares_of_two_sides):
                print(side1,side2,longest_side)
```
