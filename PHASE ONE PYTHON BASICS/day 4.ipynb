```python
squares = []
for x in range(10):
    i = x**2
    squares.append(i)
print(squares)
```

```python
squares = [x**2 for x in range(10)]
print(squares)
```

```python
is_even = [ i for i in range(10) if i % 2 == 0]
print(is_even)
```

```python
is_even = []
for i in range(10):
    if i % 2 == 0:
        is_even.append(i)
print(is_even)
```

```python
planets = ["mercury","venus","earth","mars","jupiter","saturn","uranus","neptune"]
planet_s = [32 for planet in planets]
print(planet_s)
```

```python
nums = [5,-1,-2,0,3]
def count_negatives(nums):
    n_negatives = 0
    for num in nums:
        if num < 0:
         n_negatives = n_negatives + 1
         return n_negatives
result = count_negatives(nums)
print(result)      
                                
```

```python
nums = [5,-1,-2,0,3]
def count_negatives(nums):
    return len(num for num in nums if num < 0)
print(count_negatives(nums))
```

```python
nums = [5,-1,7,0,3]
def has_lucky_number(nums):
    result = any([num % 7 == 0 for num in nums])
    return result
print(has_lucky_number(nums))
```
