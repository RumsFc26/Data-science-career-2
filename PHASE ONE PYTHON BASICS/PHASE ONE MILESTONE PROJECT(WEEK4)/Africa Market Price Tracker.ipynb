```python
#19, Analyses module
#i, we can compute the average price per product across all cities and countries in the dataset using the groupby() method in pandas. This will give us an idea of the overall price trends for each product in Africa.
average_price_per_product = Africa_market_data_df.groupby('product')['price_fcfa'].mean()

print("Average price per product across all cities and countries in the dataset:")

print(average_price_per_product)
```

```python
#ii, computing the most expensive city per product across all countries in the dataset using the groupby() method in pandas. This will give us an idea of which city has the highest prices for each product in Africa.
#idxmax() finds and return the row label of the maximum price or highest price for a product
most_expensive_city_per_product = Africa_market_data_df.loc[Africa_market_data_df.groupby("product")["price_fcfa"].idxmax()]

print("\nMost expensive city per product across all countries in the dataset:")

print(most_expensive_city_per_product[["product", "city", "country", "price_fcfa"]])
```

```python
#iii,price range per product(max-min)per product across all cities and countries in the dataset using the groupby() method in pandas. This will give us an idea of the price range for each product in Africa.
# this computes maximum and minimum prices(agg(["max","min"]))
price_range_per_product = Africa_market_data_df.groupby("product")["price_fcfa"].agg(["max", "min"])
#we calculate price range and create a column for it

price_range_per_product["Range"] = price_range_per_product["max"] - price_range_per_product["min"]

print("Price Range Per Product")
print(price_range_per_product)
```

```python
#20,functions module: write 5 reusable functions — load_data(), clean_data(), 
#compute_summary(), filter_by_city(city), and export_report(df, filename).
#1 this functions reads a csv file and returns a dataframe or it loads a csv_file
def load_data(filename):
    """
    Loads the CSV file into a pandas DataFrame.
    """

    df = pd.read_csv(filename)

    return df
df =load_data("Africa_market_data.csv")
df

```

```python
#2,clean_data()function
#this function would clean dat by removing duplicate rows,fill missing values and covert some columns like the date column in to their correct data types as shown below
def clean_data(df):
    '''
    Clean the dataset
    ''' 
#Removes duplicates rows
    df = df.drop_duplicates()  
#Filling missing values
    df["city"] = df["city"].fillna("NAN")
    df["country"] = df["country"].fillna("NAN") 
#DO SAME FOR ALL OTHER COLUMNS,ACCESS THEM AND FILL THE EMPTY FIELDS OF EACH COLUMN WITH NAN OR UNKNOWN
#Convert wrong datatypes in to correct ones example dates from int64 to  datetime64[ns] 
    df["date"] = pd.to_datetime(df["date"]) 
    df.dtypes
#you could also covert for other products suppose the datatype for the product is not correct after checking
    print("Data successfully cleaned")
    return df
df = clean_data(df)
```

```python
#3 Compute Summary() funtion
#this function performs all required anlyses
def compute_summary(df):
    print("\nAverage price per product:")
    print(df.groupby("product")["price_fcfa"].mean())
    print("\nMost expensive city per product or for each product:")
    expensive = df.loc[df.groupby("product")["price_fcfa"].idxmax(),["product","city","price_fcfa"]]
    print(expensive)
    print("\nPrice range:")
    price_range = df.groupby("product")["price_fcfa"].agg(lambda x:x.max()-x.min())
    print(price_range)
compute_summary(df)
    

```
