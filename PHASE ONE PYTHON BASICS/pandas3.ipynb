```python
#assigning your own column name,index,name to a series
import pandas as pd
s1 = pd.Series([10,20,30], index = ["a","b","c"], name ="marks")
print(s1)
```

```python
# you could create a series from a dictionary
data = {"a":"tse", "e":True, "1":3}
print(pd.Series(data))
```

```python
import numpy as np
np_array = np.array([10,20,30,40])
print(pd.Series(np_array))
#Using array to make a series
```

```python
#shows all indexes of a series or dataframe
print(s1.index)
```

# More on data frames
-a spread sheet like a google sheet or an excel sheet in pandas is known as a dataframe

-a data frame is a collection of series that contain data 

-a column in excel sheet or google sheet is called a series

```python
#creating a data frame
data = {"Names":["nde","anye","tse"],"Age":[24, 26,12], "City":["usa","new york","dehli"]}
df = pd.DataFrame(data)
df
```

```python
#  You could also make a df from a dictionary that is creating a list of multiple dictionaries
data_1 =[
{"Names":"nde","Age":24,"City":"usa"},{"Names":"anye", "Age":26, "City":"new york"},{"Names":"tse",
"Age":12,"City":"dehli"}
]
df1 = pd.DataFrame(data_1)
df1
# we can note here that data is stored as (column_name ,value),(column_name,value),and (column_name value) as the code will create a single row with all the column values

```

```python
#creating a df from an array we see that the pandas will assign indexes as show
arr_data = np.array([[1,2],[3,4],[5,6]])
df2 = pd.DataFrame(arr_data)
print(df2) 
```

```python
# From the above, you can pass the column names as A and B instead of using the default column names 0 and 1
arr_data = np.array([[1,2],[3,4],[5,6]])
df3 = pd.DataFrame(arr_data,columns = ['A','B'])
print(df3) 
```

# ACCESSING THE  of DF

```python
print(type(df1))
#getting multiple columns
print(df1[["Names","Age"]])
#making a separation between output
print("="*40)
#selecting the first row with loc
print(df1.loc[0])
print("="*40)
#selecting the second row with iloc
print(df1.iloc[1])
print("="*40)
#selecting the first two rows using slicing here, start at 0 and stop at 2 so loc do not follow the normal slicing of python.The simple reason here is that loc is using labels so will select all the rows having those labels.
print(df1.loc[0:2])
print("="*40)
#but iloc follows slicing ie start at 0, up to and not including 2
print(df1.iloc[0:2])
print("="*40)
```
