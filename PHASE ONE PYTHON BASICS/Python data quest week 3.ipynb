```python
#TASK 3(FILTERING LAB)
high_happy = df.loc[df["Score"] > 7]
high_happy
# this filter is analytically interesting because it help us isolate happiest countries and study major factors contributing to their happiness
```

```python
un_happy = df.loc[ df["Score"] < df["Score"].mean()]
un_happy
# this filter is interesting because it allows us isolate countries with low happiness levels hence allowing us identify the causes of their low happiness such as low social support and low freedom to make choices 
```

```python
average = df["Freedom to make life choices"].mean()
df.loc[df["Freedom to make life choices"] > average]
# this help us isolate the countries whose freedom to make life choices is high and thus help us compare it to happiness score thus we can determine if greater freedom to make life choices contributes tom higher satisfaction or happiness
```

```python
rich_low_happy = df.loc[(df["GDP per capita"] > df["GDP per capita"].mean()) & (df["Score"] < df["Score"].mean())]
rich_low_happy
#this filter is analytically interesting because it finds countries that should be happy but are not .that is countries that are rich but are not are not happy 
```

```python
freedom_unhappy = df[(df["Freedom to make life choices"] > df["Freedom to make life choices"].mean()) & (df["Score"] < df["Score"].mean())]
freedom_unhappy
# this filter condition is as wel interesting because it filters countries that are free but are unhappy hence showing that freedom alone is not sufficient for well-being and other economic factors must also be taken in to consideration
```

```python
# TASK 4(COLUMN CREATOR(adding 3 new columns))
#1 adding a column that is normalizing a numeric column to 0-1 range 
MIN = df["Score"].min()
MAX = df["Score"].max()
df["Normalized_Score"] = (df["Score"]-MIN)/ (MAX-MIN)
df
```

```python
#2  adding a column that is converting its country entries to upper cases
df["Capitalized_names"]= df["Country or region"].str.upper()
df
```

```python
#3 adding a boolean column flagging countries reaching some threshold
df["Rich_Countries"] = df["GDP per capita"] > df["GDP per capita"].mean() 
df
```
