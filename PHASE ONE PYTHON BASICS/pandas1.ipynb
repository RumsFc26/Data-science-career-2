```python
#pandas is a python library or package which is used to analyse and store data
#pandas works mainly with dataframes(2D)and series(1D)
#Creating Series:a pandas series is like a column in a table
# SERIES
import pandas as pd 
import numpy as np
s = pd.Series([10, 20, 30, 40]) 
print("series is :")
print(s)
#DATAFRAME: a data frame on the other hand is a 2 dimensional structure like a 2D array or a table with rows and columns
df =pd.DataFrame({"Names":["nde","rums","nsoh","ndo"], "Ages":[10, 20, 30, 40]})
print("data frame is :")
print(df)
```

```python
#Exploring a data frame 
#1 printing the first two rows of the data frame 
print(df.head(2))
#2 get every information or description about that dataframe 
print(df.info())
#3 get all statistical information about the dataframe for ONLY NUMERIC DATA  and the statistical information here is  as mean,count and min or max even
print(df.describe())
```

```python
#SELECTING AND FILTERING DATA
#1 Column selection you simply index the column label for that column(this is done using a native accessor eg simple square brackets or using the dot eg df.Names )
print(df["Names"])
#ROW SELECTION OR ACCESSING IS DONE USING THE LOC(Label based accessor) AND ILOC(Index based accessor),They are called the PANDAS ACCESSORS. WHAT THEY DO IS THEY RETURN ROW FIRST B4 COLUMN WHICH IS THE OPPOSITE OF WHAT NATIVE ACCESSORS IN PYTHON DO(the normal way of selecting or accessing data in python without using loc and iloc eg simple square brackets).the simple difference is that with the native accessors, the go to column first b4 row but with pandas accessors they go to row first b4 column
print("\n")
#2 since we know the column names we simply get the same thing 
print(df.Names)
print("\n")
#3 Row selection using the iloc(selection by position or the row number(index based selection) NB: 1 here represents the row number 1(actually the second row since numbering here starts from zero and not one) and note that row counting starts from zero)
print(df.iloc[1])
#the above will return the entire row two since iloc works as iloc[row,column] ie access the row b4 going to the column but since in this case only row was targeted, it will return only the entire row 2.we also have to take note of what the output looks like,it looks like a series that is  the row is converted in to the column found in a series and then the column labels become the indexes 
print(df.iloc[1,1])
#the above code shows how we can use the iloc to access a specific value and not and entire row
#4 Row selection using labels (selection by index, HERE WE USE THE LOC accessor )
print("\n")
print(df.loc[0])
#the above code using loc means we use the row labels of a particular row to access the row eg label for row 1 is 0, so we shall use df.loc[0]
print(df.loc[1,"Ages"])
#this above can be used to get a specific value
#CONDITION FILTERING pass the condition you want to filter as an argument in to the data frame
print(df[df['Ages'] <= 50])
# the above looks like df[...] ie like column selection but it is not what happens is that df['Ages'] <= 50] will return a column of boolean values and the df[boolean values] will return the rows for which the boolean values are true or 1
#see note book for more 
```

```python
#MODIFICATION OF A DATA FRAME
# we can add,update or drop columns and rows from a data frame
# 1 ADDING COLUMNS 
df["countries"] = ["usa", "canada", "japan","china"]
print(df)
# 2 UPDATING VALUES OF A COLUMN
df['Ages'] = df["Ages"] + 1
print(df)
#3 UPDATING A ROW VALUE
df.loc[0,'Ages'] = df.loc[0,'Ages'] + 5
#3 dropping values(drop a column) 
df.drop("countries", axis = 1, inplace =True) 
print(df)
#drop a row
df.drop(2,inplace =True)
print(df)
#note that inplace = true modifies the original dataframe while inplace = false returns the new dataframe and leaves the original one unchanged.note that for row we are to use axis = 0 but since the is already included inbuilt there is no need for it
#find out and add here

```

```python
#HANDLING MISSING VALUES
# missing values arise when we collect data and there are some times some entries are empty, most machine learning model cannot handle missing values so, there is a way we check if there are missing values and we can fill those missing values as well as we can drop them
df2 = pd.DataFrame({"A":[3, np.nan, 3], "B":[4, 5, np.nan]})
print("original:")
print(df2)
# Nan simply means empty and is a special value used to represent missing values in a dataframe or invalid numerical data. to represent a missing value in a dataframe,we use(np.nan)
#You can decide to fill the NaN values with zeroes
print("\n fill NaN with zero:") 
df3 = df2.fillna(0)
print(df3)
# Also we can decide to drop the rows having empty values or NaNs
print("\n drop rows with Nan :")
print(df2.dropna())
```

```python
#Grouping and Aggregation
#aggregation and grouping are use a lot when we want to do a lot of dat analyses using pandas data frame 
df3 = pd.DataFrame({"salary":[200, 400, 600, 800],"department":["sales", "sales", "marketing", "marketing"]})
print(df3)
print(df3.groupby("department")["salary"].mean())
# this group the dta by department and the calculates the mean salary
```
