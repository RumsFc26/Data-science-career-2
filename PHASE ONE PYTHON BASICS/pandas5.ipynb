# iterations in pandas

```python
import pandas as pd
```

```python
df = pd.DataFrame({"Name":["rums", "tse", "nde"], "Age":[25, 20, 23],"City":["Bamenda","Yaounde", "Douala"]})
df
```

```python
# we can use "iterrows()" to iterate over rows
# it will return the rows for all the columns one after another and will also return column names as indexesfor each iteration as see in the output below
for index, row in df.iterrows():
    print(row, end ="\n\n")
```

```python
# you could also print with indexes
for index, row in df.iterrows():
    print(index,row, end ="\n\n")
```

```python
# if you want all the indexes, you can use better print statements
for index, row in df.iterrows():
    print(f"{index}:{row["Name"]},{row["Age"]},{row["City"]}")
```

```python
# similar to iterrows(), we have intertuples() where by no indexes are returned but only rows eg
for row in df.itertuples():
    print(row)
```

```python
# now for elements or rows of a particular column we get
for row in df.itertuples():
    print(row.Name)
```

```python
#iterating over the indexes of the data frame
for i in df.index:
    print(i)
#df.index returns a list of all the indexes so, we are simply iterating over a list
```

```python
# iterating and print the columns of the df
for col in df.columns:
    print(col)
    print(df[col].values, end="\n\n")
# the second print statement returns all the values of  column as a list
# df.columns returns a list of all the column names, so you simply print each column name for each iteration
```

# condition based column creation using costum logic

```python
# this is mostly used in data engineering to create a column based on a condition
df.head()

```

```python
# this is a good example of feature engineering where by with the help of our existing features(columns), we are able to create  new features(columns)
values =[]
# we first create a list  or series which will store column values as shown above
# now we go to the condition as seen below
for index, row in df.iterrows():
    if row["Age"] < 30:
        values.append("young")
    else:
        values.append("senior")
df["catergory"] = values
df        

```

```python
#The vectozised or better way to do this is using apply() method as shown below. the above method is slower since we have to be looping 
df["catergory"]=df["Age"].apply(lambda age : "young" if age< 30 else "senior")
# lambda is a single line function
# this same thing could be done with interrows, intertuples and apply()
df
```

# Sort
in python we have two types of sorts whcih are 
1, sort_values: sort using values(data values)
2, sort_index: sort using indexes(rows/columns)

```python
titanic_data_set_github_url ="https://raw.githubusercontent.com/datasciencedojo/datasets/refs/heads/master/titanic.csv" 
```

```python
df = pd.read_csv(titanic_data_set_github_url)#it reads raw data from the csv file having this url,creates a data frame and then stores it in a dataframe called df
df
```

```python
# here we sort the df by age and select the first five rows of the Name and age column of the sorted DataFrame
sorted_df = df.sort_values(by="Age")
sorted_df[["Name", "Age"]].head(5)
```

```python
# selecting specific portion of your data to check ages 
sorted_df.loc[0:100, ["Name","Age"]]
```

```python
#looking at the values for age in the age column, we can see that the values of age are of many different types:float(0.45),NaN and int(22) but to solve this we have to do some analysis, we could check and correct the data type of the age column
df["Age"].dtype
```

```python
# we have to convert it to int64 that is 
df["Age"] = df["Age"].astype("int64")
```

```python
# W get error saying NaN can be converted so we have to fill empty values and the convert
df["Age"] = df["Age"].fillna(0).astype("int64")
df["Age"].dtype
```

```python
# now we got the corrected dtype and also the sorting and age values will change lets check
sorted_df = df.sort_values(by="Age")
print(sorted_df[["Name","Age"]].head(20))
```

```python
# sort by multiple columns
df_sorted = df.sort_values(by=["Age","Name"]) #Ascending order
df_sorted.head()
```

```python
# chosing the order in which you want to sort a column since by default it is always sorted in ascending order
df_sorted = df.sort_values(by=["Age","Name"],ascending=[False,True]) 
df_sorted.head() 
#here, False is for first column(Age) and True is for second column(Name) 
```

```python
# After cleaning a df, you can save it in any file format of your choice using the method. The file name will be any file format of you choice and after executing the code,a file automatically appears in your current working directory
# df_sorted.to_csv("<filename>")
df_sorted.to_csv("a.csv")
```
