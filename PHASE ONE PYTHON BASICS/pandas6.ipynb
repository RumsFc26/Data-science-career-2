```python
import pandas as pd
```

```python
#  so far we have worked with the following data types:
# Numerical data types:int&float
# - str
# - bool
# now we will look at categorical data
# it is data that consist of strings but later for machine learning it will be converted to numerical data
#Categorical data can consist of either labels(you could group them and give them a label name. so we can call them label data) or textual data
#Categorical data in Data Science and Machine Learning is data that represents groups, types, or categories rather than numerical amounts.
# examples of categorical data include:Gender → categorical data (Male, Female), Department → categorical data (Computer Science, Biology, Physics), Grade → categorical data (A, B, etc.)
```

```python
# among the columns for Name,Age,Subject and Class, subje t and class are category 
# data which have have like one million different values is not a category 
```

```python
titanic_data_set_github_url ="https://raw.githubusercontent.com/datasciencedojo/datasets/refs/heads/master/titanic.csv" 
```

```python
df = pd.read_csv(titanic_data_set_github_url)
df
```

```python
print(df["Sex"].dtype)
```

```python
# convert a column to categorical
df['Sex_C'] = df["Sex"].astype("category")
```

```python
df[["Sex", "Sex_C"]].head()
# looking at the ouput nothing has actually changed but internally,Sex_C is no longer of type "object"
```

```python
# we can see clearly that the data type of "Sex_C" is now  category hence this column could be used to explore and derive  more things as shown in the other cells below  
print(df["Sex_C"].dtype)
```

```python
#using the "cat.codes" method on the column will covert each of the categories(Male,Female) in to their individual codes(1,0).this is by default.
# cat simply means category
print(df["Sex_C"].cat.codes.head())
```

```python
df["Sex_C_Categorical_codes"] = df["Sex_C"].cat.codes
df[["Sex", "Sex_C","Sex_C_Categorical_codes"]].head()
# looking at the codes females comes first because because it categorical code compared to the index column is of lower index(also tis could also be because of an alphabetical sort done on the Sex_C column before conversion to categorical codes
```

```python
# Suppose we have a machine learning system, machine learning system does not understands string/object. it can only take numerical data.
# machine learning system can only understand 0, 1 , some number(11.2) or some very large number.
# now getting categorical codes without using cat.codes. ie using python code  do it so this pandas methods like cat.codes just make work easy
df["sex_apply_category"] = df["Sex"].apply(lambda x : 1 if x == "male" else 0) 

```

```python
# checking the unique values of this column one ca see that it is also a categorical column
df["Pclass"].unique()
```

```python
# suppose we want to do something really weired 
# firstly we will learn how to use more than one value to classify data in a column("adf","war" and "tyu") and not just using boolean(1 or 0)
df["Pclass_new_weird"] = df["Pclass"].apply(lambda x: "adf" if x == 3 else("war" if x == 2 else "tyu"))
```

```python
# now displaying it
df[["Pclass","Pclass_new_weird"]].head()
```

```python
# Suppose the data set you downloaded did not have the column for "Pclass" as numerical values but instead it had the values of the "Pclass_new_weird" column
# And from the information about the data on the internet  it is said that adf > tyu > war
# but clearly that comparison(adf > tyu > war) does not make sense for string data so we will create a categorical column and indicate the order(increasing sizes) of data so that comparison can now be possible in that column
df["Pclass_new_weird_ordered"] = pd.Categorical(
    df["Pclass_new_weird"],
    categories=["adf","tyu","war"],# our own defined order(custom(ascending))
    ordered = True    
)
```

```python
print(df["Pclass_new_weird_ordered"].dtype)
```

```python
# Let us check how many categories we have 
df["Pclass_new_weird_ordered"].cat.categories 
```

```python
# now we can use the data values to do comparison since they are no longer of type "object" but are now of type "category"
df["Pclass_new_weird_ordered"] > "tyu"
# now we can define are own costum order on words that do not make sense and compare them
```
