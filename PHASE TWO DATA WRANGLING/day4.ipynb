```python
# Missing values

#1 Detecting missing values, we use
#isnull()
#isna()
#notna(): Will return True for non-missing values


#2,Handling missing values
#i,dropna():Drops/removes rows or columns containing missing values

#ii,When should you use dropna()?
#You should not automatically delete every row containing a missing value.
# It can be reasonable when:
# only a small number of rows have missing values
# the missing rows are not important
# there is no sensible way to estimate the missing value
# the dataset is large enough that losing a few rows doesn't matter

# For example, if you have 100,000 rows and only 20 rows have missing values, dropping those 20 might be acceptable.

# But if you have 100 rows and 40 contain missing values, deleting them would be a bad idea.

#3,fillna():Instead of deleting missing values, you can replace them with something.
#When to fill missing values
#i,Fill with the Mean:When is mean useful? : Mean is useful when the data is reasonably balanced and doesn't contain strong outliers.
#ii,Fill with the Median:Median is especially useful when there are outliers or the data is skewed.
#iii,Fill with the Mode:Mode is especially useful for categorical data.
#iv,Forward Fill:This is particularly useful for time-series data, where the previous value may be a reasonable estimate.
#vi,Interpolation:Estimate a missing value using the values around it


#A we have a key word called "how" which is used with dropna().
#1,dropna(how="all"): removes an entire row only if all the values are missing
#2,dropna(how="any"): removes a row if any value is missing

#B we also have thresh which is used to specify the required minimum number of non-missing values eg dropna(thresh=2)  will keep the rows having at least two non missing values 

#C,subset:it is used to select the specific columns to look at eg df.dropna(subset=["Age"]) will remove rows where age is missing without caring if other columns have missing values 
# Duplicate detection 
#1,duplicated():it is use to determine which rows are duplicates and will return boolean
#2,drop_duplicates():it is used to remove duplicates once you have found them using duplicated()
#3,df.drop_duplicates(subset=["Name", "City"]):will allow you drop duplicates only for a given set of features or columns 
#4, you can decide which instance of the duplicate is kept when you drop duplicates 
#eg df.drop_duplicates(keep="first") or last,second or last or you could even remove all the copies of duplicates using df.drop_duplicates(keep =False).Here you also remove the main copy 
```
